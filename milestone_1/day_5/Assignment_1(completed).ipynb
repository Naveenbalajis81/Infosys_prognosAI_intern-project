#### 3. Feature Engineering – Aggregation and Rolling Statistics
Task:
- For a selected sensor (e.g., `sensor_2`), compute and append to the DataFrame:
  - The mean and standard deviation of `sensor_2` by each `engine_id`

  - The rolling mean (window size=5) for each engine’s `sensor_2` readings
- Expected Output: Show a few sample rows with these new features.

```python
sensor_cols = [col for col in df.columns if col.startswith('sensor_')]
engine_aggs=df.groupby('engine_id')[sensor_cols].agg(['mean','std'])
engine_aggs.columns=['_'.join(col) for col in engine_aggs.columns]
engine_aggs.reset_index(inplace=True)
print(f" Aggregation feature matrix shape:{engine_aggs.shape}")
engine_aggs.head()
```

Rolling Statistics

```python
for col in sensor_cols:
  df[f"{col}_rolling_mean"]=df.groupby("engine_id")[col].rolling(window=5,min_periods=1).mean().reset_index(drop=True)
  df[f"{col}_rolling_std"]=df.groupby("engine_id")[col].rolling(window=5,min_periods=1).std().reset_index(drop=True)
cols_to_show =sensor_cols+[f"{col}_rolling_mean" for col in sensor_cols]+[f"{col}_rolling_std" for col in sensor_cols]
df[cols_to_show].head(10)
```

#### 5. Rolling Window Sequence Generation
Task:
- Write a function to generate rolling window sequences of length 30 for the sensors of each engine. Save the resulting 3D NumPy array (num_sequences, window_size, num_features), and save the corresponding engine_id and cycle identifier lists as CSV files. Print the shape of the resulting sequence array and contents of the first entry.
- Expected Output: Array shape, example content, and confirmation of successful file saves.

```python
df.head()
```

```python
exclude_cols = ['engine_id','cycle']
features_cols =[col for col in df.columns if col not in exclude_cols]
df=df.sort_values(['engine_id','cycle']).reset_index(drop=True)
```

```python
def generate_rolling_windows(data,engine_col,feature,windows_size=30):
  sequence=[]
  engine_ids=[]
  cycle_ids=[]
  for engine in data[engine_col].unique():
    engine_data=data[data[engine_col]==engine]
    engine_features=engine_data[feature].values

    for i in range(windows_size-1,len(engine_data)):
      seq=engine_features[i-windows_size+1:i+1]
      sequence.append(seq)
      engine_ids.append(engine)
      cycle_ids.append(engine_data.iloc[i]['cycle'])
  sequence=np.array(sequence)
  return sequence,engine_ids,cycle_ids
```

```python
window_size=30
sequence,engine_ids,cycle_ids=generate_rolling_windows(df,'engine_id',features_cols,window_size)

print("shape of rolling window sequence:", sequence.shape)
print("Example Sequence Shape:",sequence[0].shape)
```

```python
print(f"Engine ID:{engine_ids[0]},cycle{cycle_ids[0]}")
print("Sequence data for first time windows (dhape{}):".format(sequence[0].shape))
print(sequence[0])
```

```python
assert sequence.shape[1] == window_size, "Sequence window length mismatch"

assert all(cycle_ids[i] > cycle_ids[i-1] or engine_ids[i] != engine_ids[i-1] for i in range(1, len(cycle_ids))), "Cycle order violation"

print("Basic validation checks passed.")
```

```python
np.save('rolling_window_sequences.npy', sequence)
pd.DataFrame({'engine_id': engine_ids, 'cycle': cycle_ids}).to_csv('sequence_metadata.csv', index=False)
```
